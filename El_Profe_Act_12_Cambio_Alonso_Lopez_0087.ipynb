## Las variables
1. CP --> Costo del producto
2. CD --> Cantidad de dinero
3. Cambio

## Formulas
$Cambio = CD - CP$

## Diagrama de flujo LucidChart

<img src="https://raw.githubusercontent.com/AlonsoLopezGarcia128/Calcular-Cambio-if-else-/refs/heads/main/DF%20if%20else%20cambio.png" alt="Perrito negro" width="300" height="500">



```python
# Para cuando es verdadero (si)
# Calculando el cambio usando if else
# Datos de entrada CD, CP
# Mostrar Cambio si es falso
# Si es verdadero mostrar no alcanza
CP = int(input("Ingresa el costo del producto: "))
CD = int(input("Ingresa la cantidad de dinero: "))
if CP > CD:
    print("No alcanza")
else:
    Cambio = CD - CP
    print("El cambio es: ", Cambio)
print("Programa realizado por Alonso Lopez Garcia NL = 31")
```

```python
# Para cuando es falso (no)
# Calculando el cambio usando if else
# Datos de entrada CD, CP
# Mostrar Cambio si es falso
# Si es verdadero mostrar no alcanza
CP = int(input("Ingresa el costo del producto: "))
CD = int(input("Ingresa la cantidad de dinero: "))
if CP > CD:
    print("No alcanza")
else:
    Cambio = CD - CP
    print("El cambio es: ", Cambio)
print("Programa realizado por Alonso Lopez Garcia NL = 31")
```
